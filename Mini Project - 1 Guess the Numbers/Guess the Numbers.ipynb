## Level - 1 Guess the Number by adding the looping statement

#### Using For Loop 

```python
secret_number = 56

for i in range(0,7):
    user_number = int(input("Enter a Number:"))
    
    if user_number == secret_number:
        print("Congratulations!!, You Guess Right Number")
        break
    else:
        print("Better Luck Next Time!!") 
```

#### Using While Loop

```python
secret_number = 56

while True:
    user_number = int(input("Enter a Number:"))
    
    if user_number == secret_number:
        print("Congratulations!!, You Guess Right Number")
        break
    else:
        print("Better Luck Next Time!!") 
```

## Level - 2 Guess the Number by adding the some conditions

#### Conditions are,
- User input Must between 1 to 100, otherwise shown an error message.
- If User Input is > Secret Number means shown the number is Greater than Secret number.
- If User Input is < Secret number means shown the number is lesser than Secrete number.

```python
print("Welcome to Number Guessing Game!!!")

print("Guess the number between 1 to 100")

secret_number = 56

for i in range(0,7):
    user_number = int(input("Enter a Number:"))

    if user_number < 1 or user_number > 100:
        print("You Enter Invalid Number, Please Enter Number Between 1 to 100")
        
    elif user_number > secret_number:
        print("Guess Number is Greater than Secret Number!!")
        
    elif user_number < secret_number:
        print("Guess Number is Lesser than Secret Number!!") 
        
    elif user_number == secret_number:
        print(f"Congratulations!!, You have Guessed the Number {secret_number} in {i} attemps.")
        break
        
    else:
        print("Better Luck Next Time!!") 
```

#### Using while Loop (Infinite loop)

```python
import random

print("Welcome to Number Guessing Game!!!\n")
print("Guess the number between 1 to 100\n")

secret_number = random.randint(1,100)

while True:
    user_number = int(input("Enter a Number:"))

    if user_number < 1 or user_number > 100:
        print("You Enter Invalid Number, Please Enter Number Between 1 to 100")
        
    elif user_number > secret_number:
        print("Guess Number is Higher!!")
        
    elif user_number < secret_number:
        print("Guess Number is Lesser!!") 
        
    else:
        print("Your Guessed number is correct")
        print(f"Congratulations!!, You have Guessed the Number {secret_number}.")
        break  
```

## Level - 5 Guess the Number by Handle Value Error

```python
import random

print("Welcome to Number Guessing Game!!!\n")
print("Guess the number between 1 to 100, You have only 7 Attempts\n")

secret_number = random.randint(1,100)

for i in range(0,7):

    try :
        user_number = int(input("Enter a Number:"))
    
        if user_number < 1 or user_number > 100:
            print("You Enter Invalid Number, Please Enter Number Between 1 to 100")
            
        elif user_number > secret_number:
            print("Guess Number is Higher!!")
            
        elif user_number < secret_number:
            print("Guess Number is Lesser!!") 
            
        else:
            print("Your Guessed number is correct")
            break
    except ValueError:
        print("Invalid Character, Please enter the numbers between 1 to 100 !!!")

if user_number == secret_number:
    print(f"Congratulations!!, You have Guessed the Number {secret_number} in {i} attemps.")
else:
    print("You Reached the Maximum 7 attempts of Execution, Sorry Try it again!!") 
```
